### **TO CONVERT ORDER DATE**

```python
df["order_date"]=pd.to_datetime(df["order_date"])
df
```

```python
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()
df
```

**TOTAL SALES**

```python
total_revenue = df['revenue'].sum()
print("Total Revenue:", round(total_revenue, 2))
```

**TOTAL ORDERS**

```python
total_orders = df['order_id'].nunique()
print("Total Orders:", total_orders)
```

**TOTAL QUANTITY SOLD**

```python
total_quantity = df['quantity'].sum()
print("Total Quantity Sold:", total_quantity)
```

**REVENUE BY CATEGORY**

```python
category_sales = (
    df.groupby('product_category')['revenue']
      .sum()
      .sort_values(ascending=False))
print(category_sales)
```

**VISUALIZATION**

```python
category_sales.plot(kind='bar', figsize=(8,5))
plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.xticks(rotation=0)
plt.show()
```

**REVENUE BY REGION**

```python
region_sales = (
    df.groupby('region')['revenue']
      .sum()
      .sort_values(ascending=False))
print(region_sales)
```

**VISUALIZATION**

```python
region_sales.plot(kind='bar', figsize=(8,5))
plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.xticks(rotation=0)
plt.show()
```

**MONTHLY REVENUE TREND**

```python
monthly_sales = (
    df.groupby(df['order_date'].dt.to_period('M'))['revenue']
      .sum())
monthly_sales.plot(figsize=(12,5))
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

**PAYMENT METHOD**

```python
payment_sales = (
    df.groupby('payment_method')['revenue']
      .sum()
      .sort_values(ascending=False))
print(payment_sales)
```

**PAYMENT METHOD(VISUALIZATION)**

```python
payment_sales.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(4,4))
plt.title("Revenue Distribution by Payment Method")
plt.ylabel("")
plt.show()
```

**CUSTOMER RATING**

```python
rating_by_category = (
    df.groupby('product_category')['customer_rating']
      .mean()
      .sort_values(ascending=False))
print(rating_by_category)
```

**CUSTOMER RATING(VISUALIZATION)**

```python
rating_by_category.plot(
    kind='bar',
    figsize=(8,5))
plt.title("Average Customer Rating by Category")
plt.xlabel("Product Category")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.show()
```

**DELIVERY PERFORMANCE**

```python
delivery_region = (
    df.groupby('region')['delivery_days']
      .mean()
      .sort_values())
print(delivery_region)
```

**DELIVERY PERFORMANCE(VIZUALIZATION)**

```python
delivery_region.plot(
    kind='bar',
    figsize=(8,5))
plt.title("Average Delivery Days by Region")
plt.xlabel("Region")
plt.ylabel("Average Delivery Days")
plt.show()
```

**DISCOUNT ANALYSIS**

```python
df['discount_percent']=df["discount"]*100
df
```

```python
discount_analysis=(df.groupby("discount_percent")['revenue'].sum())
print(discount_analysis)
```

**DISCOUNT VS REVENUE**

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x='discount_percent',
    y='revenue')
plt.title("Discount vs Revenue")
plt.xlabel("Discount (%)")
plt.ylabel("Revenue")
plt.show()
```

**DELIVERY DAYS VS RATING**

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,
    x='delivery_days',
    y='customer_rating')
plt.title("Delivery Days vs Customer Rating")
plt.xlabel("Delivery Days")
plt.ylabel("Customer Rating")
plt.show()
```

**CORRELATION ANALYSIS**

```python
numeric_columns = [
    'quantity',
    'unit_price',
    'discount',
    'delivery_days',
    'customer_rating',
    'revenue']
corr = df[numeric_columns].corr()
plt.figure(figsize=(7,6))
sns.heatmap(
    corr,
    annot=True,
    fmt='.2f')
plt.title("Correlation Matrix")
plt.show()
```

**TOP 10 CUSTOMERS**

```python
top_customers = (
    df.groupby('customer_id')['revenue']
      .sum()
      .sort_values(ascending=False)
      .head(10))
print(top_customers)
```

**TOP 10 CUSTOMERS(VISUALIZATION)**

```python
top_customers.sort_values().plot(
    kind='barh',
    figsize=(8,5))
plt.title("Top 10 Customers by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Customer ID")
plt.show()
```
